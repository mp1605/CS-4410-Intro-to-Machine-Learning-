**Dataset:** RDatasets copy of the `ggplot2` diamonds dataset


## Parts (a) and (b): Import pandas and load the CSV directly from a URL

`index_col=0` uses the first CSV column (the record numbers) as the DataFrame index, exactly as requested in the exercise.


```python
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline

DATA_URL = "https://vincentarelbundock.github.io/Rdatasets/csv/ggplot2/diamonds.csv"
```

```python
df = pd.read_csv(DATA_URL, index_col=0)

print(f"Dataset loaded successfully: {df.shape[0]:,} rows and {df.shape[1]} columns.")
print("Columns:", list(df.columns))
```

## Part (c): Display the first seven rows

```python
df.head(7)
```

## Part (d): Display the last seven rows

```python
df.tail(7)
```

## Part (e): Descriptive statistics for numerical columns

Calling `df.describe()` summarizes the seven numerical columns: `carat`, `depth`, `table`, `price`, `x`, `y`, and `z`.


```python
df.describe()
```

## Part (f): Descriptive statistics for categorical columns

The Series method `describe()` is called separately for `cut`, `color`, and `clarity`; the results are combined only to make the output easier to compare.


```python
categorical_summary = pd.DataFrame({
    "cut": df["cut"].describe(),
    "color": df["color"].describe(),
    "clarity": df["clarity"].describe()
})

categorical_summary
```

## Part (g): Display the unique category values

```python
for column in ["cut", "color", "clarity"]:
    print(f"{column}: {df[column].unique()}")
```

## Part (h): Histograms for all seven numerical columns

`%matplotlib inline` enables inline plots in Colab/IPython. The DataFrame `hist()` method creates one histogram for every numerical column.


```python
df.hist(figsize=(14, 10), bins=10, edgecolor="black")
plt.suptitle("Histograms of the Seven Numerical Diamond Columns", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```
